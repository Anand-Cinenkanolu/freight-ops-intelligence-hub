```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
import random
import json

np.random.seed(42)
random.seed(42)
```

```python
carriers = ['XPO Logistics', 'Estes Express', 'Old Dominion', 'ABF Freight', 'Saia LTL']
origins   = ['Chicago IL', 'Dallas TX', 'Atlanta GA', 'Los Angeles CA', 'Houston TX', 'New York NY']
dests     = ['Chicago IL', 'Dallas TX', 'Atlanta GA', 'Los Angeles CA', 'Houston TX',
             'New York NY', 'Seattle WA', 'Denver CO', 'Miami FL', 'Phoenix AZ']
svc_types = ['LTL Standard', 'LTL Expedited', 'LTL Economy', 'Volume LTL']
industries= ['Manufacturing','Retail','Healthcare','Automotive','Food & Beverage','Electronics','Chemical']
statuses  = ['Delivered','In Transit','Delayed','Cancelled','Lost']
rate_map  = {'LTL Standard':0.07,'LTL Expedited':0.14,'LTL Economy':0.05,'Volume LTL':0.04}

start_dt = datetime(2024, 1, 1)
end_dt   = datetime(2024, 12, 31)

def rdate(s, e):
    return s + timedelta(days=random.randint(0, (e - s).days))
```

```python
rows = []
for i in range(10000):
    pickup   = rdate(start_dt, end_dt)
    transit  = random.randint(1, 7)
    promised = pickup + timedelta(days=transit)
    status   = random.choices(statuses, weights=[0.82, 0.06, 0.07, 0.03, 0.02])[0]

    if status == 'Delivered':
        delay  = random.choices([-1, 0, 1, 2, 3], weights=[0.05, 0.60, 0.20, 0.10, 0.05])[0]
        actual = promised + timedelta(days=delay)
        on_time = 1 if actual <= promised else 0
    else:
        actual  = None
        on_time = 0

    carrier  = random.choice(carriers)
    origin   = random.choice(origins)
    dest     = random.choice([d for d in dests if d != origin])
    svc      = random.choice(svc_types)
    weight   = round(random.uniform(100, 10000), 1)
    distance = random.randint(100, 2500)
    cost     = round(weight * rate_map[svc] * (distance / 1000), 2)
    revenue  = round(cost * random.uniform(1.15, 1.45), 2)
    has_claim= 1 if random.random() < 0.03 else 0
    claim_amt= round(random.uniform(200, 8000), 2) if has_claim else 0.0

    rows.append({
        'shipment_id':            f'SHP{str(i+1).zfill(7)}',
        'customer_id':            f'CUST{str(random.randint(1, 800)).zfill(4)}',
        'carrier_name':           carrier,
        'service_type':           svc,
        'industry':               random.choice(industries),
        'origin_city':            origin,
        'destination_city':       dest,
        'pickup_date':            pickup.strftime('%Y-%m-%d'),
        'promised_delivery_date': promised.strftime('%Y-%m-%d'),
        'actual_delivery_date':   actual.strftime('%Y-%m-%d') if actual else None,
        'shipment_status':        status,
        'on_time_flag':           on_time,
        'weight_lbs':             weight,
        'distance_miles':         distance,
        'promised_transit_days':  transit,
        'actual_transit_days':    (actual - pickup).days if actual else None,
        'freight_cost_usd':       cost,
        'freight_revenue_usd':    revenue,
        'has_claim':              has_claim,
        'claim_amount_usd':       claim_amt
    })

df_shipments = pd.DataFrame(rows)
print(f"Generated: {len(df_shipments)} shipment records")
df_shipments.head(3)
```
